```python
def summaryFramelength(inputFile,outputFile):
    # from collections import Counter
    frame_dict = {}
    # open file
    with open(inputFile,'r') as input:
            for line in input:
                fileds = line.split()
                start_codon_pos = 51
                # Bowtie uses zero-based offset, transform to 1-based
                stop_codon_pos = int(fileds[3].split('|')[3]) + 2
                # Bowtie uses zero-based offset, transform to 1-based
                align_pos = int(fileds[4]) + 1
                length = len(fileds[5])
                # relative distance
                rel2st = align_pos - start_codon_pos
                rel2sp = align_pos - stop_codon_pos
                # assign frame
                yushu = abs(align_pos - start_codon_pos)%3
                # key
                key = '|'.join([str(length),str(yushu),str(rel2st),str(rel2sp)])
                # init dict and count
                frame_dict.setdefault(key,0)
                frame_dict[key] += 1
    # output
    output = open(outputFile,'w')
    # output
    for key,val in frame_dict.items():
        elem = key.split('|')
        output.write('\t'.join([elem[0],elem[1],elem[2],elem[3],str(val)]) + '\n')

    output.close()
```

```python
import os

# make folder
os.mkdir('./2.length-frame-data')
```

```python
sample = ['ssb1-inter-rep1.map','ssb1-inter-rep2.map','ssb2-inter-rep1.map','ssb2-inter-rep2.map',
          'ssb1-trans-rep1.map','ssb1-trans-rep2.map','ssb2-trans-rep1.map','ssb2-trans-rep2.map']

# run
for i in range(0,8):
    summaryFramelength(sample[i],''.join(['2.length-frame-data/',sample[i],'.frame.txt']))
```
